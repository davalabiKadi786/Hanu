# Task 1 — Authorized BestBuy Canada Reviews and Sentiment

Best Buy Canada's current Conditions of Use prohibit scraping and data mining. Do not automate collection unless you have written permission that specifically authorizes it. Otherwise, obtain review data through an authorized route and run the sentiment workflow on that export. Never bypass CAPTCHA, bot controls, IP restrictions, or other access controls.

```python
from pathlib import Path
import subprocess, sys

PRODUCT_URL = 'PASTE_BESTBUY_CANADA_PRODUCT_URL_HERE'
PERMISSION_DOCUMENT = Path('path/to/written_permission.pdf')
MAX_PAGES_PER_SORT = 5
OUTPUT = Path('outputs/bestbuy_reviews.csv')
if not PERMISSION_DOCUMENT.is_file():
    raise PermissionError('Do not automate collection without written permission from Best Buy.')
if 'PASTE_' in PRODUCT_URL:
    raise ValueError('Set PRODUCT_URL only after written permission authorizes collection.')
subprocess.run([sys.executable, '-m', 'src.bestbuy_scraper', '--url', PRODUCT_URL, '--pages', str(MAX_PAGES_PER_SORT), '--output', str(OUTPUT), '--permission-document', str(PERMISSION_DOCUMENT)], check=True)
```

## Sentiment approach
The collector runs VADER on the combined title and body, then assigns simple aspect tags with polarity suffixes. This lexicon baseline should be validated against a hand-labeled sample; it is not a fine-tuned transformer model. Sorting and pagination rely on current public-page controls and should be checked in the exported CSV.

```python
import pandas as pd
reviews = pd.read_csv(OUTPUT)
print('Rows:', len(reviews), 'Unique IDs:', reviews['id'].nunique())
display(reviews.head())
display(reviews['sentiment'].value_counts(dropna=False))
display(reviews['aspect_sentiment_tags'].value_counts().head(20))
```
